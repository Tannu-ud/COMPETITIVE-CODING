```python
def containsNearbyDuplicate(nums, k):
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] == nums[j] and abs(i - j) <= k:
                return True
    return False
k = int(input("Enter k: "))
nums = list(map(int, input("Enter array elements: ").split()))
print(containsNearbyDuplicate(nums, k))
```

```python
#hash map
def containsNearbyDuplicate(nums, k):
    seen = {}
    for i in range(len(nums)):
        if nums[i] in seen and i - seen[nums[i]] <= k:
            return True
        seen[nums[i]] = i
    return False
k = int(input("Enter k: "))
nums = list(map(int, input("Enter array elements: ").split()))
print(containsNearbyDuplicate(nums, k))
```

```python
#sliding window
def containsNearbyDuplicate(nums, k):
    window = set()
    for i in range(len(nums)):
        if nums[i] in window:
            return True
        window.add(nums[i])
        if len(window) > k:
            window.remove(nums[i - k])
    return False
k = int(input("Enter k: "))
nums = list(map(int, input("Enter array elements: ").split()))
print(containsNearbyDuplicate(nums, k))
```
